```python
import librosa
from librosa import display
import matplotlib.pyplot as plt
import numpy as np
# Load the desired audio files
y, sr = librosa.load('../data/genres_original/rock/rock.00000.wav', duration=30)
y2, sr2 = librosa.load('../data/genres_original/metal/metal.00000.wav', duration=30)
# Compute a mel-scaled power spectrogram
S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128, fmax=8000)
S2 = librosa.feature.melspectrogram(y=y2, sr=sr2, n_mels=128, fmax=8000)
# Convert to log scale (dB). We'll use the peak power as reference.
S_dB = librosa.power_to_db(S, ref=np.max)
S2_dB = librosa.power_to_db(S2, ref=np.max)

# Plot the spectrogram
plt.figure(figsize=(10, 4))
display.specshow(S_dB, x_axis='time', y_axis='mel', sr=sr, fmax=8000)
plt.colorbar(format='%+2.0f dB')
plt.title('Mel Spectrogram')
plt.tight_layout()
plt.show()

# Plot the second spectrogram
plt.figure(figsize=(10, 4))
display.specshow(S2_dB, x_axis='time', y_axis='mel', sr=sr2, fmax=8000)
plt.colorbar(format='%+2.0f dB')
plt.title('Mel Spectrogram')
plt.tight_layout()
plt.show()

```
